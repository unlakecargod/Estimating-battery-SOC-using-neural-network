```python
import vsigmoid
import vtanh
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x = np.linspace(-10, 10, num=500).reshape(-1, 1)
```

```python
def sig(x):
	return 1/(1 + np.exp(-x))
```

```python
def tanh(X):
    return (np.exp(X)-np.exp(-X))/(np.exp(X)+np.exp(-X))
```

```python
y1 = vsigmoid.hardsigmoid2(x)
```

```python
y2 = sig(x)
```

```python
y3 = vtanh.hardtanh2(x)
```

```python
y4 = tanh(x)
```

```python
plt.plot(y1)
plt.plot(y2)
plt.show()
```

```python
plt.plot(y3)
plt.plot(y4)
plt.show()
```

```python
data = pd.DataFrame(np.concatenate((x, y1, y2, y3, y4), axis=1))
```
