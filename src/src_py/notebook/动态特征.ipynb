功能描述：  
输入ui与soc_ini，生成安时积分的soc_pre以及对应的模型参数。  
输入单位为mV和mA，时间单位为s。

以下为模块的主体部分

```python
def ah(soc_ini, i, cap):
    soc_pre = soc_ini - i / (cap * 3600)
    return soc_pre
```

```python
def dfgen(input_dict):
    """加载电流电压，初始soc以及参数查找表三个变量"""
    ui = input_dict['ui']
    soc_ini = input_dict['soc_ini']
    param = input_dict['param']
    """逐时间步生成特征"""
    df = []
    for i in range(ui.shape[0]):
        soc_pre = ah(soc_ini, ui[i, 0], cap)
        index = int(np.round(soc_pre * 10))
        feature = param[index]
        features = np.concatenate((ui[i, :], soc_pre.reshape(-1), feature))
        df.append(features)
    df_out = np.array(df)
    return df_out
```
